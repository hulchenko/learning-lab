To demonstrate your familiarity with OpenAI API, and also Ollama, build a tool that takes a technical question,  
and responds with an explanation. This is a tool that you will be able to use yourself during the course!

```python
# imports
import os
from openai import OpenAI
from dotenv import load_dotenv
from IPython.display import Markdown, display, update_display
```

```python
# constants

# GPT
MODEL_GPT = 'gpt-4o-mini'
# base_url is not needed for OpenAI (default)

# LLAMA
MODEL_LLAMA = 'llama3.2'
base_url = 'http://localhost:11434/v1'
```

```python
# set up environment
load_dotenv(override=True)
api_key = os.getenv('OPENAI_API_KEY')

system_prompt = """
You are a knowledgeable assistant that is proficient in software development and Python specifically.
You are given a question or a question with a code snippet, and your response must be in ASD-STE100 format, one paragraph only, no lists, maximum 100 words.
If it's a python code, you must provide a small example in a separate code block that will complement the explanation.
You must also create a 3 question quiz that will test the user's understanding of the code or its relevant concepts.
"""
```

```python
# here is the question; type over this to ask something new
def get_question(question):
    return f"""
        Explain the following: {question}
        """

# question = get_question(f"""
# what is a difference between array and a tuple in python? how do I manipulate them?
# """)

question = get_question(f"""
sliced_tuple = my_tuple[1:3]
""")

request = [
    {"role": "system", "content": system_prompt},
    {"role": "user", "content": question},
]
```

```python
# Get gpt-4o-mini to answer, with streaming
model_client = OpenAI(
    # grabs OPENAI_API_KEY from .env for api_key
    # does not need base_url
)
response = model_client.chat.completions.create(model=MODEL_GPT,messages=request)

display(Markdown(response.choices[0].message.content))
```

```python
# Get Llama 3.2 to answer
model_client = OpenAI(
    base_url=base_url,
    api_key="test" # can be any
)

response = model_client.chat.completions.create(model=MODEL_LLAMA,messages=request)
display(Markdown(response.choices[0].message.content))
```
