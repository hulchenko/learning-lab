## Do you know what an Endpoint is?

If not, please review the Technical Foundations guide in the guides folder

And, here is an endpoint that might interest you...

````python
import requests

headers = {"Authorization": f"Bearer {api_key}", "Content-Type": "application/json"}

payload = {
    "model": "gpt-6-luna",
    "reasoning_effort": "none",
    "messages": [
        {"role": "user", "content": "Tell me a fun fact"}]
}

payload
````

````python
response = requests.post(
    "https://api.openai.com/v1/chat/completions",
    headers=headers,
    json=payload
)

response.json()
````

````python
response.json()["choices"][0]["message"]["content"]
````

# What is the openai package?

It's known as a Python Client Library.

It's nothing more than a wrapper around making this exact call to the http endpoint.

It just allows you to work with nice Python code instead of messing around with janky json objects.

But that's it. It's open-source and lightweight. Some people think it contains OpenAI model code - it doesn't!


````python
# Create OpenAI client

from openai import OpenAI
openai = OpenAI()

response = openai.chat.completions.create(model="gpt-6-luna", reasoning_effort="none", messages=[{"role": "user", "content": "Tell me a fun fact"}])

response.choices[0].message.content

````

## And then this great thing happened:

OpenAI's Chat Completions API was so popular, that the other model providers created endpoints that are identical.

They are known as the "OpenAI Compatible Endpoints".

For example, google made one here: https://generativelanguage.googleapis.com/v1beta/openai/

And OpenAI decided to be kind: they said, hey, you can just use the same client library that we made for GPT. We'll allow you to specify a different endpoint URL and a different key, to use another provider.

So you can use:

```python
gemini = OpenAI(base_url="https://generativelanguage.googleapis.com/v1beta/openai/", api_key="AIz....")
gemini.chat.completions.create(...)
```

And to be clear - even though OpenAI is in the code, we're only using this lightweight python client library to call the endpoint - there's no OpenAI model involved here.

If you're confused, please review Guide 9 in the Guides folder!

And now let's try it!

````python
gemini = OpenAI(base_url=GEMINI_BASE_URL, api_key=google_api_key)

response = gemini.chat.completions.create(model="gemini-3.1-flash-lite", messages=[{"role": "user", "content": "Tell me a fun fact"}])

response.choices[0].message.content
````

````python
# for api key and such (not needed for local Ollama models)
# import os
# from typing import override
# from dotenv import load_dotenv

# constants
OLLAMA_BASE_URL = "http://localhost:11434/v1"
LLM_MODEL = "llama3.2"
urlList = ["https://www.bbc.com", "https://news.yahoo.co.jp", "https://www.globo.com"]
system_prompt = """
You are a media person, working in entertainment, that analyzes the contents of a website, and provides the summary of only one, the most interesting article, found across multiple popular news websites.
Response must be concise, in markdown and in English only. Translate if needed. Do not wrap the markdown in a code block - respond just with the markdown.
"""
user_prompt = f"""
What's going on in the world nowadays?
"""

# some helpful tools
from scraper import fetch_website_contents
from IPython.display import Markdown, display

# open AI client library in python
# this is just a wrapper for HTTP method to connect to LLM models
# this does NOT mean it's for open AI only
from openai import OpenAI

def messages_for(news_summary):
    return [
        {"role": "system", "content": system_prompt + news_summary},
        {"role": "user", "content": user_prompt}
    ]

def summarize(urls):
    parts = []
    for url in urls:
        parts.append(f"Website: {url}\n{fetch_website_contents(url)}")
    website_contents_stringified = "\n\n".join(parts)

    response = ollama.chat.completions.create(
        model = LLM_MODEL,
        messages = messages_for(website_contents_stringified)
    )
    return response.choices[0].message.content

def display_summary(urls):
    summary = summarize(urls)
    display(Markdown(summary))

ollama = OpenAI(base_url=OLLAMA_BASE_URL)
display_summary(urlList)
````
